# Bollinger Bands Trading Strategy

US Equity : Netflix 
Period : 01/01/2026-10/01/2026




```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import yfinance as yf
```

```python
start_date = "2026-01-01"
end_date = "2026-10-01"


data = yf.download("NFLX", start=start_date, end=end_date, auto_adjust=True, threads=True, progress=False)['Close']
```

```python
data.tail()
```

```python
data.plot(label="Netflix", figsize=(14,8))
```

```python
mavg_20 = data['NFLX'].rolling(20).mean()
rstd_20 = data['NFLX'].rolling(20).std()


data['Bollinger Upper'] = mavg_20 + 2*rstd_20
data['Bollinger Lower'] = mavg_20 - 2*rstd_20

```

```python
data.tail()
```

```python
data['NFLX'].plot(label='NFLX', figsize=(16,8))
data['Bollinger Upper'].plot(label='Bollinger Upper Bound')
data['Bollinger Lower'].plot(label='Bollinger Lower Bound')
plt.legend()
```

```python
dernier_mois = data.loc['2026-09']
dernier_mois['NFLX'].plot(label='Netflix', figsize=(16, 8))
dernier_mois['Bollinger Upper'].plot(label='Bollinger Upper Bound')
dernier_mois['Bollinger Lower'].plot(label='Bollinger Lower Bound')
plt.legend()
```
